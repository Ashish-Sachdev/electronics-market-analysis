# Electronics Market Dataset Exploration

## 1. Overview
This notebook loads the raw electronics pricing dataset, checks data quality, and removes unusable columns before further analysis.

## 2. Data loading
The first step is to read the source CSV and inspect the dataset structure.


```python
import pandas as pd

# Load the raw dataset
csv_path = '../data/raw/products_pricing_data.csv'
df = pd.read_csv(csv_path)

# Display the dataset shape and basic structure
print(f"The dataset has {df.shape[0]} rows and {df.shape[1]} columns.")
print('\nDataset preview:')
display(df.head())

# Basic information about each column
print('\nDataset information:')
df.info()

# Show the column names
print('\nColumns in the dataset:')
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")
```

```python
## 3. Cleaning the dataset
# Remove columns with 'Unnamed' in the name
# These are usually metadata artifacts from Excel exports
cleaned_df = df.loc[:, ~df.columns.str.contains('Unnamed', case=False, na=False)]

# Drop columns with more than 90% missing values
missing_threshold = 0.90
cleaned_df = cleaned_df.loc[:, cleaned_df.isna().mean() < missing_threshold]

# Optional validation
print(f"After cleaning, the dataset has {cleaned_df.shape[0]} rows and {cleaned_df.shape[1]} columns.")
print('\nRemaining columns:')
for i, col in enumerate(cleaned_df.columns, start=1):
    print(f"{i}. {col}")

# Keep the cleaned dataset for downstream analysis
df = cleaned_df
```

```python
## 4. Data quality checks
# Check for duplicate records
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates}")

# Optional: inspect duplicates if any
if duplicates > 0:
    display(df[df.duplicated(keep=False)].head())

# Summary of missing values
print('\nMissing values by column:')
missing_values = df.isna().sum().sort_values(ascending=False)
display(missing_values[missing_values > 0])

## 5. Save cleaned dataset
# Save the cleaned dataset to the interim folder.
# This preserves the original raw file and keeps a reusable cleaned version for later steps.
output_path = '../data/interim/cleaned_empty_columns_products_pricing_data.csv'
df.to_csv(output_path, index=False)
print(f"Cleaned dataset saved to: {output_path}")
```
